## 16. Correlation analysis

Select numerical variables:

```python
numeric_cols = [
    'Cumulative',
    'Cumulative  Test positive',
    'Cumulative  tests performed',
    'Discharged',
    'Expired',
    'Home Quarantine',
    'New  (last 24 hrs)',
    'Still admitted',
    'Tests  performed in last 24 hours'
]

# Clean the 'Tests  performed in last 24 hours' column
df['Tests  performed in last 24 hours'] = pd.to_numeric(
    df['Tests  performed in last 24 hours'].astype(str).str.replace('No ', ''), 
    errors='coerce'
)

corr = df[numeric_cols].corr()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix')
plt.show()
```

## 18. Important derived metrics

For this dataset, I recommend creating these variables:

```python
df['Case Fatality Rate'] = (
    df['Expired'] /
    df['Cumulative  Test positive']
) * 100

df[['Date', 'Region', 'Expired', 'Cumulative  Test positive', 'Case Fatality Rate']].head(10)
```

```python
df['Recovery Rate'] = (
    df['Discharged'] /
    df['Cumulative  Test positive']
) * 100

df['Recovery Rate'] = df['Recovery Rate'].round(2)

df[['Date', 'Region', 'Discharged', 'Cumulative  Test positive', 'Recovery Rate']].head(10)
```

```python
df['Positivity Rate'] = (
    df['Cumulative  Test positive'] /
    df['Cumulative  tests performed']
) * 100

df['Positivity Rate'] = df['Positivity Rate'].round(2)

df[['Date', 'Region',
    'Cumulative  Test positive',
    'Cumulative  tests performed',
    'Positivity Rate']].head(10)
```

```python
df.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    df['Date'],
    df['Case Fatality Rate'],
    label='Case Fatality Rate'
)

plt.plot(
    df['Date'],
    df['Recovery Rate'],
    label='Recovery Rate'
)

plt.plot(
    df['Date'],
    df['Positivity Rate'],
    label='Positivity Rate'
)

plt.title('COVID-19 Key Indicators Over Time')
plt.xlabel('Date')
plt.ylabel('Rate (%)')
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```
