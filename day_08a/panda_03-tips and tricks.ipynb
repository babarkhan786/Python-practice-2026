# 16. Spliting a string into multiple columns

```python
import pandas as pd
df=pd.DataFrame({'name':['Muhammmad Khan', 'Ahmad Hassan', 'Sajjad Rana', 'Ali Khan'],
                                    'Location':['Lahore, Pakistan','Multan, Pakistan','Karachi, Pakistan','Faisalabad, Pakistan']})
df
```

```python
# spliting a column into two columns
# adding new columns
df[["first_name", "last_name"]] = df.name.str.split(' ', expand=True)
df
```

```python
df[["City", "Country"]] = df.Location.str.split(', ', expand=True)
df
```

```python
#refine data manipulation
df = df[['first_name', 'last_name','City','Country']]
df
```

## 23. Reshaping a dataframe

```python
fasla = pd.DataFrame([['2345', 100, 200, 300], ['34567', 400, 500, 600],['67890', 700, 800, 900]],
                     columns=['zip', 'factory', 'warehouse', 'retail']
)

fasla.head()
```

```python
sns.barplot(x="zip", y="factory",data=fasla)
```

```python
fasla.head().T
```

```python
fasla_long = fasla.melt(
    id_vars="zip",
    var_name="location_type",
    value_name="distance"
)
```
