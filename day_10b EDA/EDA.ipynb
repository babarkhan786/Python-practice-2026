## 11. Outlier analysis

```python
#Fare:
sns.boxplot(data=kashti, x="fare")
plt.title("Fare Outliers")
plt.show()
```

```python
#Age:
sns.boxplot(data=kashti, x="age")
plt.title("Age Outliers")
plt.show()
```

```python
#Calculate IQR:
Q1 = kashti["fare"].quantile(0.25)
Q3 = kashti["fare"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = kashti[
    (kashti["fare"] < lower) |
    (kashti["fare"] > upper)
]

outliers
```

## 12. Feature relationships

```python
#Did women survive more than men?
pd.crosstab(
    kashti["sex"],
    kashti["survived"],
    normalize="index"
)
```

```python
#Did first-class passengers survive more?
pd.crosstab(
    kashti["pclass"],
    kashti["survived"],
    normalize="index"
)
```

```python
# Did children have better survival?
kashti["Age_Group"] = pd.cut(
    kashti["age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teenager", "Young Adult", "Adult", "Senior"]
)
```

```python
kashti.groupby("Age_Group", observed=True)["survived"].mean()
```

## Add a Name column for all 891 rows

```python
def create_name(row):
    if row["age"] < 14:
        title = "Master"
    elif row["sex"] == "female":
        title = "Miss"
    else:
        title = "Mr"
    
    return f"{title}"
```

```python
kashti["Name"] = kashti.apply(create_name, axis=1)
```

```python
kashti[["Name", "sex", "age"]].head(10)
```

```python
kashti["Title"] = kashti["Name"].str.extract(
    r",\s*([^.]+)\."
)[0]
```

```python
kashti[["Name", "Title"]].head(10)
```

```python
kashti["Title"].value_counts()
```

## 14. Missing-value treatment

```python
kashti["age"] = kashti["age"].fillna(kashti["age"].median())
```

```python
kashti["embarked"] = kashti["embarked"].fillna(
    kashti["embarked"].mode()[0]
)
```

```python
kashti.columns
```

```python
kashti["CabinKnown"] = kashti["deck"].notna().astype(int)
```

```python
kashti[["deck", "CabinKnown"]].head(10)
```

```python
kashti["CabinKnown"].value_counts()
```

```python
sns.countplot(
    data=kashti,
    x="CabinKnown",
    hue="survived"
)

plt.title("Survival by Cabin Information")
plt.xlabel("Cabin Known (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()
```
