```python
#load dataset to work with pandas
# Seaborn has data sets
import pandas as pd
import seaborn as sns
# titanic data
kashti= sns.load_dataset('titanic')
kashti.head()
```

```python
kashti.info() # rows are lso known as instences
```

```python
# what every column means here
kashti.columns
```

| Column            | Explanation                                                | Example                                  |
| ----------------- | ---------------------------------------------------------- | ---------------------------------------- |
| **`survived`**    | Whether the passenger survived                             | `1` = survived, `0` = died               |
| **`pclass`**      | Passenger's ticket class                                   | `1` = First, `2` = Second, `3` = Third   |
| **`sex`**         | Passenger's sex                                            | `male`, `female`                         |
| **`age`**         | Passenger's age in years                                   | `22.0`, `38.0`                           |
| **`sibsp`**       | Number of siblings or spouses traveling with the passenger | `0`, `1`, `2`                            |
| **`parch`**       | Number of parents or children traveling with the passenger | `0`, `1`, `2`                            |
| **`fare`**        | Price of the passenger's ticket                            | `7.25`, `71.28`                          |
| **`embarked`**    | Port where the passenger boarded                           | `C`, `Q`, `S`                            |
| **`class`**       | Passenger class as text                                    | `First`, `Second`, `Third`               |
| **`who`**         | Passenger category                                         | `man`, `woman`, `child`                  |
| **`adult_male`**  | Whether the passenger was an adult male                    | `True`, `False`                          |
| **`deck`**        | Deck where the passenger's cabin was located               | `A`, `B`, `C`, etc.                      |
| **`embark_town`** | Full name of boarding town                                 | `Cherbourg`, `Queenstown`, `Southampton` |
| **`alive`**       | Survival status as text                                    | `yes`, `no`                              |
| **`alone`**       | Whether the passenger was traveling alone                  | `True`, `False`                          |
