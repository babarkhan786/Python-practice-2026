## Git + GitHub setup for Python Practice 2026
### Run these commands in the VS Code PowerShell terminal/

```python
# 1. Go to your project folder
cd "E:\Python\Python practice 2026"
```

```python
# 2. Check Git
git --version
```

```python
# 3. Configure Git identity
git config --global user.name "Allah Ditta Babar"
git config --global user.email "YOUR_GITHUB_EMAIL"
```

```python
# Check configuration
git config --global --list
```

```python
# 4. Initialize Git
git init
git branch -M main
```

```python
# 5. Connect local project to GitHub
git remote add origin https://github.com/babarkhan786/Python-practice-2026.git
```

### If you get: "remote origin already exists", use:
# git remote set-url origin https://github.com/babarkhan786/Python-practice-2026.git

```python
# Check remote connection
git remote -v
```

```python
# 6. Create .gitignore
@"
.vscode/
__pycache__/
*.pyc
.ipynb_checkpoints/
.venv/
venv/
~$*.xlsx
"@ | Out-File -Encoding utf8 .gitignore
```

```python
# Check .gitignore
Get-Content .gitignore
```

```python

# 8. Check project status
git status

# 9. Add all project files
git add .
```

```python
# Check staged files
git status

# 10. Create the first commit
git commit -m "Initial Python practice project"

# 11. Push the project to GitHub
git push -u origin main
```

```python
# ---------------------------------------------------------
# FUTURE UPDATES
# ---------------------------------------------------------
# After modifying or adding notebooks, Python files, datasets, etc.:

git add .
git commit -m "Updated Python practice"
git push


# Examples of commit messages:
# git commit -m "Added pandas practice"
# git commit -m "Added NumPy practice"
# git commit -m "Added Seaborn visualization"
# git commit -m "Added Titanic analysis"
# git commit -m "Added Iris analysis"
```

```python
# ---------------------------------------------------------
# USEFUL GIT COMMANDS
# ---------------------------------------------------------

# See current status
git status

# See connected GitHub repository
git remote -v

# See commit history
git log --oneline

# See current branch
git branch

# Download changes from GitHub
git pull

# See files tracked by Git
git ls-files
```
