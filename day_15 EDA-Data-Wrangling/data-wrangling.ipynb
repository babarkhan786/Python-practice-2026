# Data Wrangling
data wrangling is the process of cleaning, transforming, and structuring raw table-like datasets into a usable format for analysis or modeling. This step is critical in data science workflows because messy data can lead to inaccurate insights and poor decision-making.

A typical wrangling workflow involves exploration, cleaning, transformation, validation, and publishing of the dataset. In Python, the pandas library is the most common tool for these tasks, while in R, data.table is widely used for high-performance manipulation.

# Removal of outliers
1. visualization
2. IQR method
3. Z-score

```python
# 1. Visualization
import seaborn as sns

sns.boxplot(data=df, x='age')
```

```python
# 2. IQR method
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
IQR
```

```python
# 1. remove outliers using IQR
# Short version
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1

df = df[df['age'].between(Q1 - 1.5 * IQR, Q3 + 1.5 * IQR)]

# # 2. long version
# Q1 = df['age'].quantile(0.25)
# Q3 = df['age'].quantile(0.75)

# IQR = Q3 - Q1

# lower = Q1 - 1.5 * IQR
# upper = Q3 + 1.5 * IQR

# df = df[(df['age'] >= lower) & (df['age'] <= upper)]
```

# Log transformation
For the Titanic dataset, log transformation is mainly useful for numerical variables that are right-skewed, especially Fare. Age is less strongly skewed, so transforming it is optional.

```python
kashti = sns.load_dataset('titanic')
```

```python
kashti.shape
```

```python
kashti.info()
```

```python
kashti.head()
```

```python
# Step 1: Import libraries
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Step 2: Handle missing values
kashti['age'] = kashti['age'].fillna(kashti['age'].median())
kashti['fare'] = kashti['fare'].fillna(kashti['fare'].median())
```

```python
# Step 3: Check skewness before transformation
print("Age skewness:", kashti['age'].skew())
print("Fare skewness:", kashti['fare'].skew())
# Step 4: Plot skewness before transformation
# Age
sns.histplot(data=kashti, x='age', kde=True)
plt.title('Age Distribution Before Log Transformation')
plt.show()
# Fare
sns.histplot(data=kashti, x='fare', kde=True)
plt.title('Fare Distribution Before Log Transformation')
plt.show()
```

```python
# Step 5: Box plots before transformation
# Age
sns.boxplot(data=kashti, x='sex', y='age', hue='sex')
plt.title('Age Box Plot Before Log Transformation')
plt.show()
# Fare
sns.boxplot(data=kashti, x='sex', y='fare', hue='sex')
plt.title('Fare Box Plot Before Log Transformation')
plt.show()
```

```python
# Step 6: Apply log transformation
kashti['age_log'] = np.log(kashti['age'])
kashti['fare_log'] = np.log(kashti['fare'])
```

```python
# Step 8: Plot skewness after transformation
# Age
sns.histplot(data=kashti, x='age_log')
plt.title('Age Distribution After Log Transformation')
plt.show()
# Fare
sns.histplot(data=kashti, x='fare_log')
plt.title('Fare Distribution After Log Transformation')
plt.show()
```

```python
# Step 9: Box plots after transformation
# Age
sns.boxplot(data=kashti, x='sex', y='age_log', hue='sex')
plt.title('Age Box Plot After Log Transformation')
plt.show()

# Fare
sns.boxplot(data=kashti, x='sex', y='fare_log', hue='sex')
plt.title('Fare Box Plot After Log Transformation')
plt.show()
```
