## how to deal missing values in EDA
### Categorical/object, Numerics (float, int), Boolean

```python
print("mean =", df["age"].mean())
print("median =", df["age"].median())
print("mode =", df["age"].mode())
```

### 1. Numerics (float, int)

```python
# Check missing values
df.isnull().sum()
```
