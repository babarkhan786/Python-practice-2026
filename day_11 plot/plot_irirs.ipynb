## 1. Best for correlation analysis

If you want to see relationships among numeric variables:

```python
corr = phool.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

## 2. Better-looking correlation heatmap

```python
corr = phool.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='RdYlBu_r',
    fmt='.2f',
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Correlation Matrix")
plt.show()
```

## 3. Show only the upper/lower triangle

```python
corr = phool.select_dtypes(include='number').corr()

mask = np.tril(np.ones_like(corr, dtype=bool))

sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Upper Triangle Correlation Heatmap")
plt.show()
```

```python
corr = phool.select_dtypes(include='number').corr()

mask = np.triu(np.ones_like(corr, dtype=bool))

sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

## 4. Heatmap with no numbers

```python
sns.heatmap(
    corr,
    cmap='coolwarm',
    center=0,
    square=True
)

plt.show()
```

## Recommended heatmap

For your Python/Seaborn EDA practice, use this:

```python
corr = phool.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

## Basic violin plot

```python
sns.violinplot(
    data=phool,
    x='species',
    y='sepal_length'
)

plt.show()
```

```python
#with-different-colors
sns.violinplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    legend=False
)

plt.show()
```

## Violin + individual observations

This is a very useful EDA combination:

```python
sns.violinplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    inner='box',
    legend=False
)

sns.stripplot(
    data=phool,
    x='species',
    y='sepal_length',
    color='black',
    jitter=True,
    alpha=0.4
)

plt.show()
```

```python
sns.violinplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='Set2',
    inner='quart',
    linewidth=1,
    legend=False
)
```

## 1. Box plot

Good for median, quartiles, and outliers.

```python
sns.boxplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    legend=False
)
plt.show()
```

## 2. Strip plot

Good for seeing every individual observation.

```python
sns.stripplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    jitter=True,
    legend=False
)
plt.show()
```

## 3. Swarm plot

Similar to strip plot, but Seaborn intelligently positions points to reduce overlap.

```python
sns.swarmplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    legend=False
)
plt.show()
```

## 4. Box + Strip plot

One of my recommended choices for EDA because you get summary statistics + individual data points.

```python
sns.boxplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    legend=False
)

sns.stripplot(
    data=phool,
    x='species',
    y='sepal_length',
    color='black',
    jitter=True,
    alpha=0.5
)

plt.show()
```

## 5. Box + Swarm plot

Even clearer when the dataset isn't too large.

```python
sns.boxplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    legend=False
)

sns.swarmplot(
    data=phool,
    x='species',
    y='sepal_length',
    color='black',
    size=3
)

plt.show()
```

## 6. Bar plot

Useful when you mainly want to compare mean sepal length.

```python
sns.barplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    errorbar='sd',
    legend=False
)

plt.show()
```

## 7. Point plot

Good for comparing group means and uncertainty.

```python
sns.pointplot(
    data=phool,
    x='species',
    y='sepal_length',
    hue='species',
    palette='pastel',
    errorbar='sd',
    legend=False
)

plt.show()
```

## 8. Histogram by species

```python
sns.histplot(
    data=phool,
    x='sepal_length',
    hue='species',
    palette='pastel',
    kde=True
)

plt.show()
```

## 9. KDE plot

```python
sns.kdeplot(
    data=phool,
    x='sepal_length',
    hue='species',
    fill=True,
    palette='pastel'
)

plt.show()
```
