```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import data set
kashti = sns.load_dataset('titanic')
kashti.head()
```

```python
kashti.info()
```

```python
kashti.isnull().sum()
```

```python
# how to deal null values
kashti.dropna(inplace=True)
```

```python
kashti.survived.value_counts().plot(kind='bar')
```

```python
sns.countplot(x='survived', hue='sex', data=kashti)
```

```python
# sns.scatterplot(x='age', y='fare', hue='sex', data=kashti)
sns.scatterplot(x='age', y='fare', hue='pclass', data=kashti)
```

```python
sns.barplot(x='pclass', y='fare', hue='pclass', data=kashti)
```

```python
sns.barplot(x='pclass', y='fare', hue='sex', data=kashti)
```

## How to change change the x and y axis of the plot

just change the name at x and y in code

```python
g = sns.FacetGrid(
    kashti,
    hue='survived',
    col='pclass',
    margin_titles=True,
    palette={1: 'seagreen', 0: 'gray'}
)
g = g.map(plt.scatter, "fare", "age", edgecolor="w").add_legend();
```

```python
g = sns.FacetGrid(
    kashti,
    hue='survived',
    col='sex',
    margin_titles=True,
    palette={1: 'seagreen', 0: 'gray'}
)
g = g.map(plt.scatter, "fare", "age", edgecolor="w").add_legend();
```

```python
g = sns.FacetGrid(kashti,hue='survived',col='who',margin_titles=True,
    palette={1: 'seagreen', 0: 'gray'})
g = g.map(plt.scatter, "fare", "age", edgecolor="w").add_legend();
```

```python
corr = kashti.select_dtypes(include=['number']).corr()
plt.figure(figsize=(10, 10))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
```
