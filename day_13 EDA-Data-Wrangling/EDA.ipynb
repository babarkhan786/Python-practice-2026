# EDA

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# import dataset
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
#summarize the data, summary statistics of the data
df.describe()
```

```python
# find missing or null values in the dataset boolean values, True if null else False
df.isnull()
```

```python
# null values in the dataset, count of null values in each column
df.isnull().sum()
```

```python
#percentage of null values in the dataset, percentage of null values in each column
df.isnull().sum() / len(df) * 100
```

```python
# find the number of unique values in each column
# to know variables that are categorical and numerical
# categorical variables have less unique values than numerical variables

df.nunique()
```

```python
df.columns
```

```python
# visulize the nan values in the dataset using heatmap
sns.heatmap(df.isnull())
```

```python
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
# cbar is the color bar on the right side of the heatmap, wheras the cmap is the color map of the heatmap, which is the color of the heatmap, and yticklabels is the y axis labels
sns.heatmap(df.isnull(), yticklabels=False, cbar=True, cmap='viridis')
```

```python
# remove deck column from the dataset, as it has more than 70% null values
df.drop(columns='deck', inplace=True)
```
