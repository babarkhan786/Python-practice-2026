```python
# ============================================
# COVID-19 Key Indicators EDA
# Case Fatality Rate, Recovery Rate & Positivity Rate
# ============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Make sure Date is in datetime format

df = pd.read_csv("covid-19-cases-in-pakistan-2020.csv")

# --------------------------------------------
# 1. Calculate Case Fatality Rate
# --------------------------------------------

df['Case Fatality Rate'] = (
    df['Expired'] /
    df['Cumulative  Test positive']
) * 100


# --------------------------------------------
# 2. Calculate Recovery Rate
# --------------------------------------------

df['Recovery Rate'] = (
    df['Discharged'] /
    df['Cumulative  Test positive']
) * 100


# --------------------------------------------
# 3. Calculate Positivity Rate
# --------------------------------------------

df['Positivity Rate'] = (
    df['Cumulative  Test positive'] /
    df['Cumulative  tests performed']
) * 100


# --------------------------------------------
# 4. Round indicators to 2 decimal places
# --------------------------------------------

df['Case Fatality Rate'] = df['Case Fatality Rate'].round(2)
df['Recovery Rate'] = df['Recovery Rate'].round(2)
df['Positivity Rate'] = df['Positivity Rate'].round(2)


# --------------------------------------------
# 5. Display calculated indicators
# --------------------------------------------

print("COVID-19 Key Indicators:")

display(
    df[
        [
            'Date',
            'Region',
            'Case Fatality Rate',
            'Recovery Rate',
            'Positivity Rate'
        ]
    ].head(10)
)


# ============================================
# 6. Plot Case Fatality Rate
# ============================================

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=df,
    x='Date',
    y='Case Fatality Rate'
)

plt.title('Case Fatality Rate Over Time')
plt.xlabel('Date')
plt.ylabel('Case Fatality Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================
# 7. Plot Recovery Rate
# ============================================

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=df,
    x='Date',
    y='Recovery Rate'
)

plt.title('Recovery Rate Over Time')
plt.xlabel('Date')
plt.ylabel('Recovery Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================
# 8. Plot Positivity Rate
# ============================================

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=df,
    x='Date',
    y='Positivity Rate'
)

plt.title('COVID-19 Positivity Rate Over Time')
plt.xlabel('Date')
plt.ylabel('Positivity Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================
# 9. Combined Comparison of Three Indicators
# ============================================

plt.figure(figsize=(14, 6))

plt.plot(
    df['Date'],
    df['Case Fatality Rate'],
    label='Case Fatality Rate'
)

plt.plot(
    df['Date'],
    df['Recovery Rate'],
    label='Recovery Rate'
)

plt.plot(
    df['Date'],
    df['Positivity Rate'],
    label='Positivity Rate'
)

plt.title('COVID-19 Key Indicators Over Time')
plt.xlabel('Date')
plt.ylabel('Rate (%)')

plt.legend()
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```
